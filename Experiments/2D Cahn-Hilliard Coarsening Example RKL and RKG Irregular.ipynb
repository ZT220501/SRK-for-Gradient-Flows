```python
# Time and space range
a_x = 0
b_x = 2 * np.pi
a_y = 0
b_y = 2 * np.pi

#Initial x grid size and t grid size
Nx = 256
Ny = 256

dx = float64((b_x - a_x) / Nx)
dy = float64((b_y - a_y) / Ny)

T = 2
eps = 0.1

dt = 1e-2

bc= "Neumann"
X, Y, mask = utils.L_shaped_domain(Nx, a_x, b_x, a_y, b_y)

save = True
save_path = "../../Results/Cahn_Hilliard_Irregular"


print("Initial dt used is " + str(dt))
print("eps used is " + str(eps))
print("32 * dt / dx^2 is " + str(32 * dt / dx**4))
```

```python
'''
Initial Condition for the Allen-Cahn simulation
'''
def initial_condition(mesh_grid):
    X, Y = mesh_grid
    np.random.seed(12345)
    return 0.0025 + 0.004 * np.random.uniform(-1, 1, size=X.shape).astype(np.float64)
```

```python
my_solver_rkl = solvers.Cahn_Hilliard_Solver_2D_Constant_Timestep_Irregular_Domain("RKL2_2D_CH_irregular", 200, dt, X, Y, mask, initial_condition, T, eps, bc=bc, energy_case=3, plot_frequency=10, splitting_order="NLN", save=save, save_path=save_path)
my_solver_rkg = solvers.Cahn_Hilliard_Solver_2D_Constant_Timestep_Irregular_Domain("RKG2_2D_CH_irregular", 240, dt, X, Y, mask, initial_condition, T, eps, bc=bc, energy_case=3, plot_frequency=10, splitting_order="NLN", save=save, save_path=save_path)
    
# my_solver.initialize(display=True)
my_solver_rkl.initialize(display=True)
my_solver_rkg.initialize(display=True)
```

```python
my_solver_rkl.solve()
```

```python
my_solver_rkg.solve()
```

```python
u_total_rkl, t_total_rkl, energy_total_rkl, mesh_grid_rkl = my_solver_rkl.get_results()
u_total_rkg, t_total_rkg, energy_total_rkg, mesh_grid_rkg = my_solver_rkg.get_results()
```

```python
energy_gradient_total_rkl = energy_total_rkl[:, 0]
energy_non_linear_total_rkl = energy_total_rkl[:, 1]
energy_rkl = energy_gradient_total_rkl + energy_non_linear_total_rkl

# Journal of Computational Physics (Elsevier): Times New Roman for figures
plt.rcParams["mathtext.fontset"] = "stix"
plt.rcParams["mathtext.rm"] = "Times New Roman"

fig, ax = plt.subplots(figsize=(6, 5), dpi=120)
ax.plot(t_total_rkl[:], energy_rkl[:] * dx**2, color="#29cfcc", lw=2.5)

from matplotlib.ticker import FuncFormatter, MultipleLocator

config_title = f"{save_path}/RKL2_2D_CH_irregular/RKL2_2D_CH_irregular_CH_comparison_Nx{Nx}_Ny{Ny}_dt{dt}_eps{eps:.4f}_T{T}_bc{bc}_energy_case3"

label_font = {"fontsize": 20, "fontfamily": "Times New Roman"}
ax.tick_params(axis="both", labelsize=20)
for label in ax.get_xticklabels() + ax.get_yticklabels():
    label.set_fontfamily("Times New Roman")
# ax.legend(loc="best", frameon=True, prop={"family": "serif", "size": 14})
# ax.grid(True, linestyle="--", alpha=0.6)
ax.set_axisbelow(True)
# ax.set_yticks([6, 8, 10, 12])
fig.tight_layout()

if save:
    energy_save_path = f"{config_title}.pdf"
    fig.savefig(energy_save_path, bbox_inches="tight")
    print(f"Figure saved to {energy_save_path}")
plt.show()
```
